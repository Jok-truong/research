## **#1 :What are Numpy and Pandas?**

- **NumPy** và **Pandas** là các thư viện nền tảng trong hệ sinh thái Python, được thiết kế đặc biệt cho việc xử lý data số và thực hiện các tác vụ phân tích data. Chúng hoạt động như bộ tool chuyên biệt mục đích để làm việc với các **collections of numbers**, **tables of data**, **time series information**, và phổ biến trong các ứng dụng **data science** và **artificial intelligence**.


###  **NumPy: The Foundation for Numerical Computing**

- **NumPy**, viết tắt của **Numerical Python**, là thư viện nền tảng cho hầu hết công cụ tính toán khoa học trong Python.
- `ndarray` được coi là quan trọng nhất của **NumPy**, một kiểu `array` mạnh mẽ có thể làm việc với data một chiều, hai chiều hoặc nhiều chiều.

- **Array** trong **NumPy** trở nên đặc biệt vì:
  
  - **Efficiency**:
    - **Array** trong **NumPy** được xây dựng bằng `C` và được tối ưu mạnh cho các phép tính số học.
    - Việc tính toán theo từng phần tử trên **Array** trong **NumPy** nhanh hơn rất nhiều so với việc tính toán trên `list` của Python. Ưu điểm về tốc độ này đặc biệt quan trọng khi làm việc với các `datasets` siêu lớn.
  
  - **Memory Usage**:
    - **Array** trong **NumPy** tiết kiệm bộ nhớ hơn so với `list` trong Python, đặc biệt khi xử lý các `datasets` số lớn. Lý do là vì **NumPy** lưu trữ các phần tử cùng **data type** trong một vùng nhớ liên tục.

  - **Functionality**:
    - **NumPy** cung cấp một tập hợp đầy đủ các functions tính toán cao cấp, làm cho việc áp dụng tính toán lên `array` dễ dàng hơn.
    - Các functions cho  **linear algebra, Fourier transforms, random number generation, statistical analysis**.

- Nói một cách đơn giản nhất, bất cứ khi nào bạn cần thực hiện các phép toán trên những khối `data` số như **vectors, matrices, higher-dimensional tensors**, **NumPy** sẽ cung cấp các **core data structures và functions** để làm việc đó một cách hiệu quả.

###  **Pandas: Data Analysis and Manipulation Made Easier**

- Trong khi **NumPy** tập trung vào các phép toán số ở mức nền tảng thấp, **Pandas** lại cung cấp các **data structures** và **analysis tools**, giúp việc sử dụng trở nên thuận lợi và trực quan hơn, đặc biệt khi ta làm việc với data dạng table như **Excel** hoặc **SQL tables**.

- Có hai **data structures** chính trong **Pandas**:

  * **Series**: Là một mảng một chiều có nhãn `index`, tương tự như một cột trong bảng tính hoặc một vector dữ liệu có chỉ mục. **Series** có thể chứa dữ liệu thuộc bất kỳ kiểu dữ liệu nào của **NumPy**.

  * **DataFrame**: Là một cấu trúc dữ liệu hai chiều có nhãn với các hàng và cột. Mỗi cột có thể chứa kiểu dữ liệu khác nhau. **DataFrame** giống như một bảng Excel, một **SQL table** hoặc một tập hợp nhiều **Series**, và đây là đối tượng được sử dụng nhiều nhất trong **Pandas**

- Mối quan hệ Series ↔ DataFrame
    <pre>
    DataFrame
     ├── Series (column 1)
     ├── Series (column 2)
     └── Series (column 3)
    </pre>

```python
import pandas as pd

ages = pd.Series([25, 30, 28], index=["Alice", "Bob", "Charlie"])
print(ages)
```

```python
data = {
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 28],
    "score": [80, 90, 85]
}

df = pd.DataFrame(data)
print(df)
```

- Pandas đặc biệt mạnh trong các lĩnh vực sau:
  
  - Xử lý dữ liệu **(Data Handling)**: Dễ dàng đọc dữ liệu từ nhiều định dạng khác nhau như CSV, Excel, JSON hoặc cơ sở dữ liệu SQL, và ghi dữ liệu đã xử lý ra file.

  - Làm sạch dữ liệu **(Data Cleaning)**: Cung cấp các công cụ để phát hiện, điền hoặc loại bỏ dữ liệu bị thiếu `NaN`, lọc các hàng không cần thiết và biến đổi dữ liệu.
  
  -  **Data Exploration and Analysis**: Cho phép chọn các phần dữ liệu con, nhóm dữ liệu theo điều kiện, thực hiện các phép tính và tổng hợp (như tổng, trung bình, đếm), cũng như gộp hoặc nối nhiều tập dữ liệu lại với nhau.

  - Chuỗi thời gian **(Time Series)**: Cung cấp các công cụ chuyên biệt để làm việc với dữ liệu có yếu tố thời gian.
