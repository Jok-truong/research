##  **#1: Strong Information: Variables**



### **What is a Variable?**

Trong Python, **variable** là một tên tượng trưng (nhãn) được dùng để tham chiếu đến một vị trí trong bộ nhớ máy tính, nơi lưu trữ các giá trị (dữ liệu). Chúng giống như những gói hàng có nhãn, giúp lập trình viên dễ dàng quản lý, truy cập và thao tác với dữ liệu (int, string, list..).

Hãy tưởng tượng bạn đang sắp xếp các mặt hàng trong hộp. Bạn sẽ không chỉ đặt mọi thứ vào những hộp không được đánh dấu, bạn sẽ gắn nhãn cho chúng ('Quần áo mùa đông', 'sách',...). Các **variable** phục vụ mục đích tương tự trong lập trình: Gán tên mô tả cho các bộ nhớ chứa dữ liệu.


### **Creating Variables: The Assignment Statement**

Việc tạo một **variable** trong Python vô cùng đơn giản sử dụng **asignment operator**, ký hiệu là `=`:

`variable_name = value`

```python
message = 'Hello, Python'
# message is a variable storing a string of text
count = 10
# count is a variable storing an integer
price = 49.99
# price is a variable storing a floating-point number
is_active = True
# is_active is a variable storing a boolean value
```

Vì Python là **dynamically typed**, có nghĩa là bạn không cần phải khai báo rõ ràng **type of data** cho **variable**. Python sẽ tự động xác định type dựa trên **value** mà bạn gán cho nó.

### **Naming Variables: Rules and Conventions**

Việc đặt tên **variable** là rất quan trọng để đánh giá lập trình viên, nó giúp người đọc code dễ hiểu và dễ bảo trì. Python đã cung cấp các quy tắc và quy ước nhằm cho **naming variables**:

  * **variable** không được giống với [Python keywords](https://www.w3schools.com/python/python_ref_keywords.asp).

        * VD: Nên viết `user_name` thay vì `un`, hoặc `total_price` thay vì `tp`.

Các tên ngắn như `x, y hoặc i` có thể chấp nhận được trong một số trường hợp đặc biệt, chẳng hạn như **variable** đếm trong vòng lặp hoặc tọa độ toán học, nhưng nhìn chung vẫn nên ưu tiên các tên **variable** mang tính mô tả.

### **Changing Variable Values (Reassignment)**

Ta có thể thay đổi giá trị được lưu trữ trong **variable** bằng cách gán lại giá trị mới cho **variable** đó bằng toán tử gán `=`.

```python
# Assign an initial value
score = 100
print("Initial score:", score) # Output: Initial score: 100

# Update the value
score = 150
print("Updated score:", score) # Output: Updated score: 150

# Assign a value based on the current value
score = score + 20 # Read the current value (150), add 20, store the result (170) back
print("Final score:", score)   # Output: Final score: 170
```

##  **#2: Numbers (Integers, Floats)**


### **Integers (int)**

**Integers** đại diện cho số nguyên trong python. Chúng có thể là dương, âm hoặc 0, nhưng chúng không thể là phân số hoặc thập phân.

**Integers** có thể lớn tùy ý, nó chỉ bị giới hạn bởi bộ nhớ của hệ thống. Bạn không cần phải lo lắng về `size` như `short`, `int`, `long` ở các ngôn ngữ lập trình khác.

Bạn cũng có thể kiểm trả **data types** với `type()`

```python
# Examples of integers
user_count = 150
temperature_celsius = -5
year = 2024
zero_value = 0

print(user_count)
print(temperature_celsius)
print(type(year))
print(type(zero_value))
```

### **Floating-Point Numbers (float)**

**Floating-point numbers** hay **float** là kiểu dữ liệu dùng để biểu diễn các số thập phân (ví dụ: 3.14, -0.5, 2.0).

```python
# Examples of floating-point numbers
pi_approx = 3.14159
price = 49.99
temperature_fahrenheit = -20.5
account_balance = 0.0 # Using a decimal makes it a float
distance_meters = 1.5e3 # Scientific notation for 1.5 * 10^3, which is 1500.0

print(pi_approx)
print(distance_meters)

print(type(price))
print(type(temperature_fahrenheit))
print(type(account_balance))
```

### **0.1 + 0.2 = ?**

- Điều quan trong cần lưu ý là **floating-point numbers** trên máy tính không phải lúc nào cũng hoàn toàn chính xác. Do cách lưu trữ **floating-point numbers** nội bộ, thường lưu ở định dạng `nhị phân IEEE 754`, các phép tính liên quan đến **float** đôi khi có thể mang lại kết quả rất khác so với những gì bạn mong đợi về mặt toán học.

  -  Vd: `0.1 + 0.2` có thể dẫn đến kết quả `0.30000000000000004` thay vì `0.3`

- Trong hầu hết các trường hợp, sự sai lệch rất nhỏ này không gây ảnh hưởng gì đáng kể. Tuy nhiên, bạn cần chú ý trong các bài toán tài chính hoặc mô phỏng khoa học, nơi yêu cầu độ chính xác cao.

- Để khắc phục vấn đề này, python cung cấp `Decimal` để xử lý các trường hợp tương tự.

```python
result = 0.1 + 0.2
print(result) # output: 0.30000000000000004
```

```python
from decimal import Decimal

result = Decimal('0.1') + Decimal('0.2')
print(result) # output: 0.3
```

## **#3: Text (Strings)**

- **Text (String)** là kiểu dữ liệu cơ bản dùng để biểu diễn văn bản, là một dãy các ký tự (chữ cái, số, ký tự đặc biệt) được đặt trong cặp dấu nháy đơn `( ' ' )`, nháy kép `( " " )` hoặc ba dấu nháy `( ''' ''' hoặc """ """ )`.

```python
# Using single quotes
message_single = 'Hello, Python learner!'
print(message_single)

# Using double quotes
message_double = "This also works just fine."
print(message_double)

# What if your string contains a single quote? Use double quotes to enclose it.
quote = "He said, 'Python is fun!'"
print(quote)

# What if your string contains a double quote? Use single quotes.
reply = 'She responded, "Indeed it is."'
print(reply)

# Triple double quotes for multi-line string
multi_line_doc = """This is the first line.
This is the second line.
And "quotes" or 'apostrophes' can be used freely inside."""
print(multi_line_doc)

# Triple single quotes work the same way
multi_line_alt = '''Another way to write
multi-line text,
very convenient.'''
print(multi_line_alt)
```

### **Basic String Operations**
Bạn có thể kết hợp các chuỗi **(combine strings)** bằng toán tử **concatenation `+`** và **repetition `*`**

```python
first_name = "Ada"
last_name = "Lovelace"
space = " "

# Concatenation
full_name = first_name + space + last_name
print(full_name) # Output: Ada Lovelace

# Repetition
separator = "-" * 10 # Repeat the hyphen 10 times
print(separator)   # Output: ----------
print(first_name * 3) # Output: AdaAdaAda
```

Lưu ý về bạn **concatenation** string này trực tiếp với con những con số, thì chương trình sẽ bị lỗi. Trước tiên, bạn cần chuyển đổi số thành một string, sử dụng str(), mà chúng ta sẽ sớm đề cập trong **Type Conversion**.

### **Accessing Characters: Indexing**

Vì string là những sequences, bạn có thể truy cập các ký tự riêng lẽ bẳng cách sử dụng chỉ mục **(index)** của chúng. Python được thiết lập **index** bắt đầu từ con số `0`, nghĩa là `index` đầu tiên sẽ là `0`.

```python
language = "Python"

# Accessing characters by positive index
print(language[0]) # Output: P (the first character)
print(language[1]) # Output: y (the second character)
print(language[5]) # Output: n (the sixth character)

# Accessing characters using negative index (starts from the end)
print(language[-1]) # Output: n (the last character)
print(language[-2]) # Output: o (the second-to-last character)
```

```python
language = "Python"

# Trying to access an index that doesn't exist will result in an IndexError.
print(language[6])
```

### **Extracting Substrings: Slicing**

Nếu bạn cần lấy nhiều hơn một ký tự, bạn có thể trích xuất một phần của string, gọi là **substring** hoặc **slice**.

Thao tác này được gọi là **slicing** và sử dụng cú pháp `[start:stop:step]`.

- `string[start : end : step]`

  - `start`: vị trí bắt đầu của **slice**.  Nếu bỏ qua, mặc định là `0`.
  - `end`: Vị trí kết thúc của **slice**. Nếu bỏ qua, mặc định là đến cuối string.
  - `step`: Bước nhảy (số ký tự bỏ qua). Nếu bỏ qua, mặc định là `1`.

```python
language = "Python"

# Get characters from index 1 up to (but not including) index 4
print(language[1:4]) # Output: yth

# Get characters from the beginning up to index 3 (exclusive)
print(language[:3])  # Output: Pyt

# Get characters from index 2 to the end
print(language[2:])  # Output: thon

# Get the whole string (less common, but shows the defaults)
print(language[:])   # Output: Python

# Get every second character
print(language[::2]) # Output: Pto

# Get the string in reverse
print(language[::-1]) # Output: nohtyP
```

**Slicing** luôn tạo ra một string mới và không bao giờ gây ra lỗi `IndexError`, ngay cả khi các chỉ số bắt đầu hoặc kết thúc vượt quá độ dài của string.
Thay vào đó, Python sẽ chỉ trả về phần string có thể lấy được trong phạm vi được yêu cầu.

```python
s = "Python"

print(s[0:100])   # Python
print(s[3:50])    # hon
print(s[-100:3])  # Pyt
```

### **String Immutability**

- Một đặc điểm quan trọng của string trong Python là chúng bất biến **(immutable)**.
Điều này có nghĩa là một khi string đã được tạo ra, bạn không thể thay đổi trực tiếp nội dung của nó.

- Những thao tác trông giống như đang sửa string, chẳng hạn như **concatenation** và **repetition** hoặc các phương thức xử lý string, thực chất là tạo ra một string mới và trả về string đó, chứ không làm thay đổi string ban đầu.

```python
greeting = "hello"
# greeting[0] = "H" # This will cause a TypeError: 'str' object does not support item assignment

# To "change" the greeting, create a new string
concatenation = "H" + greeting[1:]
repetition = "H" * 3
print(concatenation) # Output: Hello
print(repetition)     # Output: hello (original string is unchanged)
```

### **Formatting Strings**

- Thông thường, bạn sẽ cần tạo các string có chứa giá trị của biến.
Mặc dù có thể dùng **concatenation** (+), cách này nhanh chóng trở nên rườm rà và dễ gây lỗi, đặc biệt khi làm việc với các kiểu dữ liệu không phải là string.

- Python cung cấp những cách tốt hơn và gọn gàng hơn để định dạng string.
Cách hiện đại và được khuyến khích nhất là sử dụng `f-string`, được giới thiệu từ Python 3.6.

- Để tạo `f-string`, bạn chỉ cần thêm chữ `f` hoặc `F` trước string.
Bên trong string, bạn có thể chèn trực tiếp biến hoặc biểu thức bằng cách đặt chúng trong dấu ngoặc nhọn `{}`.



```python
name = "Alice"
age = 30
city = "New York"

# Using f-string for clean formatting
intro_fstring = f"My name is {name}, I am {age} years old, and I live in {city}."
print(intro_fstring)
# Output: My name is Alice, I am 30 years old, and I live in New York.

# You can also include expressions inside the braces
radius = 5
area = 3.14159 * radius**2
print(f"A circle with radius {radius} has an area of {area:.2f}.")
# Output: A circle with radius 5 has an area of 31.42.
# Note: :.2f formats the float to 2 decimal places.
```

- `f-string` ngắn gọn và giúp ta dễ đọc code, nó được sử dụng để thay thể các phương thức format cũ khác như `str.format()` hoặc toán tử `%` ở các phiên bản Python cũ

```python
name = "Alice"
age = 25

print("My name is %s and I am %d years old." % (name, age)) # %s -> string, %d -> int
print("My name is {} and I am {} years old.".format(name, age))
```

### **Escape Sequences**

Đôi khi, bạn cần đưa vào string những ký tự đặc biệt - những ký tự khó hoặc không thể gõ trực tiếp, hoặc những ký tự đã có ý nghĩa đặc biệt trong Python.
Ví dụ, bạn muốn chèn ký tự xuống dòng, tab, hoặc thêm dấu ngoặc kép vào bên trong một string đang dùng chính loại ngoặc đó.

Trong Python, việc này được thực hiện bằng **escape sequence**, bắt đầu bằng dấu gạch chéo ngược `\`.

- Các **escape sequences** thường gặp:
| Escape | Ý nghĩa              |
| ------ | -------------------- |
| `\n`   | Xuống dòng (newline) |
| `\t`   | Tab                  |
| `\\`   | Dấu `\`              |
| `\'`   | Dấu nháy đơn `'`     |
| `\"`   | Dấu nháy kép `"`     |

```python
# Newline
print("First line.\nSecond line.")
# Output:
# First line.
# Second line.

# Tab
print("Column1\tColumn2")
# Output: Column1    Column2

# Literal backslash
print("This is a path: C:\\Users\\Name")
# Output: This is a path: C:\Users\Name

# Literal quotes within string defined by same quotes
print('He said, \'Hello!\'') # Output: He said, 'Hello!'
print("She replied, \"Hi!\"") # Output: She replied, "Hi!"
```

### **Common String Methods**

- String trong Python có rất nhiều phương thức **(methods)** dựng sẵn để thực hiện các thao tác phổ biến.

- **Method** giống như hàm **(function)**, nhưng được gọi trực tiếp trên một đối tượng bằng `.`.Ở đây, đối tượng chính là string.

Dưới đây là một số **functions** và **method** string thường dùng:

| Method / Function | Mô tả (English)               | Mô tả (Tiếng Việt)          | Ví dụ                      |
| ----------------- | ----------------------------- | --------------------------- | -------------------------- |
| `len(string)`          | Returns number of characters  | Trả về số ký tự trong chuỗi | `len("Hello") → 5`         |
| `s.lower()`       | Convert to lowercase          | Chuyển sang chữ thường      | `"Hi".lower() → "hi"`      |
| `s.upper()`       | Convert to uppercase          | Chuyển sang chữ hoa         | `"hi".upper() → "HI"`      |
| `s.strip()`       | Remove whitespace (both ends) | Xóa khoảng trắng đầu & cuối | `" hi ".strip() → "hi"`    |
| `s.lstrip()`      | Remove left whitespace        | Xóa khoảng trắng bên trái   | `" hi".lstrip()`           |
| `s.rstrip()`      | Remove right whitespace       | Xóa khoảng trắng bên phải   | `"hi ".rstrip()`           |
| `s.find(substring)`     | Find substring index          | Tìm vị trí chuỗi con        | `"hello".find("e") → 1`    |
| `s.replace(a, b)` | Replace substring             | Thay chuỗi con              | `"a-b".replace("-", "_")`  |
| `s.startswith(prefix)` | Check prefix                  | Kiểm tra bắt đầu bằng       | `"hi.py".startswith("hi")` |
| `s.endswith(suffix)`   | Check suffix                  | Kiểm tra kết thúc bằng      | `"hi.py".endswith(".py")`  |
| `s.split(separator)`    | Split into list               | Tách chuỗi thành list       | `"a,b".split(",")`         |
| `sep.join(iterable)`    | Join iterable                 | Nối list thành chuỗi        | `" ".join(["I","AI"])`     |


```python
text = "  Learning Python is Fun!  "

# Length
print(len(text)) # Output: 27

# Case conversion
print(text.lower()) # Output:   learning python is fun!
print(text.upper()) # Output:   LEARNING PYTHON IS FUN!

# Stripping whitespace
print(text.strip())  # Output: Learning Python is Fun!
print(text.lstrip()) # Output: Learning Python is Fun!
print(text.rstrip()) # Output:   Learning Python is Fun!

# Finding substrings
print(text.find("Python")) # Output: 11 (index where 'Python' starts)
print(text.find("Java"))   # Output: -1 (not found)

# Replacing substrings
print(text.replace("Fun", "Awesome")) # Output:   Learning Python is Awesome!

# Checking start/end (use strip first for accuracy here)
clean_text = text.strip()
print(clean_text.startswith("Learning")) # Output: True
print(clean_text.endswith("!"))        # Output: True

# Splitting a string into a list
words = clean_text.split(" ")
print(words) # Output: ['Learning', 'Python', 'is', 'Fun!']

# Joining a list into a string
joined_string = "---".join(words)
print(joined_string) # Output: Learning---Python---is---Fun!

# Original string remains unchanged due to immutability
print(text) # Output:   Learning Python is Fun!
```

## **#4: Booleans**





### **The Boolean Type: True and False**

- Kiểu `boolean` được đặt theo tên của George Boole, là 1 trong types of data trong Python. Nó chỉ có hai giá trị: `True` và `False`

```python
is_learning = True
is_finished = False

print(is_learning)
print(is_finished)
```

### **Booleans from Comparisons**

- Mặc dugf bạn có thể gắn `True` hoặc `False` trực tiếp, nhưng các giá trị `boolean` thường phát sinh do kết quả của việc so sánh. Khi sử dụng các toán tử so sánh **(comparison operators)**: `==, !=, >, <, >=, <=` kết quả luôn là giá trị `boolean`.

```python
age = 20
is_adult = age >= 18  # Is age greater than or equal to 18?
print(is_adult)      # Output: True

temperature = 15.5
is_cold = temperature < 10.0 # Is temperature less than 10.0?
print(is_cold)       # Output: False

name = "Alice"
is_bob = name == "Bob" # Is the name equal to "Bob"?
print(is_bob)          # Output: False
```

### **Booleans and Logical Operations**

- Giá trị `boolean` là nên tẳng cho việc ra quyết định trong Python. Chúng được dụng rộng rãi với các toán tử logic **(logical opertors)**: `and, or, not, xor` để kết hợp nhiều điều kiện.

```python
logged_in = True
is_admin = False

# 'and' requires both sides to be True for the result to be True
can_access_admin_panel = logged_in and is_admin
print(can_access_admin_panel) # Output: False

# 'or' requires at least one side to be True for the result to be True
can_view_content = logged_in or is_admin
print(can_view_content)       # Output: True

# 'not' inverts the boolean value
is_guest = not logged_in
print(is_guest)               # Output: False
```

### **Truthiness: When Other Types Act Like Booleans**

- Python có một khái niệm gọi là **truthiness**.
Khi Python cần một giá trị đúng/sai (ví dụ trong câu lệnh `if`), nó có thể tự động đánh giá các kiểu dữ liệu khác như là `True` hoặc `False`.

- Những giá trị được xem là `True` được gọi là `truthy`,
còn những giá trị được xem là `False` được gọi là `falsy`.

- Trong Python, một số giá trị được coi là `falsy`, nghĩa là chúng hoạt động giống như `False` khi được dùng trong các biểu thức điều kiện.

- Những giá trị `falsy` đó bao gồm:
  - Giá trị `False`
  - `None` đại diện cho việc không có giá trị
  - `0` ở mọi dạng: `0, 0.0`
  - Bất kỳ chuỗi hoặc tập hợp rỗng: `'', [], (), {}, set()`

- **Truthiness** giúp bạn viết các câu điều kiện dễ đọc và ngắn gọn hơn:
  - vd: Thay vì kiểm tra một `list` có rỗng bằng `if len(my_list) > 0 `, bạn chỉ cần `if my_list`

```python
items = []
if items: # items is empty, so it's falsy
    print("There are items in the list.")
else:
    print("The list is empty.") # This will be printed

user_name = "Charlie"
if user_name: # user_name is not empty, so it's truthy
    print(f"Hello, {user_name}!") # This will be printed
else:
    print("User name is missing.")

count = 0
if count: # count is 0, so it's falsy
    print("Count is positive.")
else:
    print("Count is zero.") # This will be printed
```

Mặc dù `True` và `False` là các giá trị `boolean` rõ ràng trong Python, việc hiểu **truthiness** giúp bạn hiểu cách Python đưa ra quyết định trong các câu lệnh điều kiện.

Cơ chế đánh giá `boolean` ngầm này giúp code Python ngắn gọn và dễ đọc hơn.

Boolean tuy đơn giản nhưng lại vô cùng quan trọng.
Chúng cho phép chương trình biểu diễn trạng thái (đúng/sai), đánh giá điều kiện thông qua phép so sánh và các **logical operators**, và từ đó đưa ra quyết định.

Vì vậy, `boolean` là nền tảng của logic lập trình, và bạn sẽ sử dụng chúng liên tục khi xây dựng các chương trình phản ứng khác nhau dựa trên dữ liệu đầu vào hoặc trạng thái bên trong.

- Lưu ý: khi trong biểu thức có toán tử logic thì toán tử này sẽ được thực hiện cuối cùng và kết quả khi nào trả về là `Boolean`


```python
result = (2 + 3) * 4
print(result) # Output: 20 (Parentheses force addition first: 5 * 4)
```

```python
calculation = 100 - 20 / 5 * 2 + 3**2
# Step-by-step evaluation:
# 1. Exponentiation: 3**2 becomes 9
#    Expression: 100 - 20 / 5 * 2 + 9
# 2. Division/Multiplication (from left to right):
#    20 / 5 becomes 4.0
#    Expression: 100 - 4.0 * 2 + 9
#    4.0 * 2 becomes 8.0
#    Expression: 100 - 8.0 + 9
# 3. Addition/Subtraction (from left to right):
#    100 - 8.0 becomes 92.0
#    Expression: 92.0 + 9
#    92.0 + 9 becomes 101.0
print(calculation) # Output: 101.0
```

## **#6: Working with Operators: Comparison**

- Khi lập trình, chúng ta thường phải so sánh các giá trị với nhau để đưa ra quyết định. Giống như dùng các phép tính cộng `+` trừ `-` để tính toán, Python có những toán tử so sánh **comparison operators** (còn gọi là **relational operators**) để kiểm tra mối quan hệ giữa hai giá trị. Những phép so sánh này rất quan trọng vì chúng giúp chương trình “quyết định” chạy theo hướng nào tùy theo điều kiện.

- Kết quả của mọi phép so sánh luôn là một giá trị **Boolean**: chỉ có thể là `True` hoặc `False`. Bạn cứ tưởng tượng như đang hỏi Python một câu hỏi có/không về hai giá trị vậy.

### **Equal To: `==`**

- Toán tử này kiểm tra xem hai giá trị có giống hệt nhau không.

```python
# Comparing numbers
print(5 == 5)
print(10 == 7)

# Comparing strings
print("hello" == "hello")
print("Python" == "python") # Case matters!

# Assigning the result to a variable
are_numbers_equal = (100 == 100)
print(are_numbers_equal)
```

### **Not Equal To: `!=`**

- Ngược lại với `==`, toán tử này kiểm tra xem hai giá trị có khác nhau không

```python
print(5 != 5)         # False
print(10 != 7)        # True

print("hello" != "")  # True
print("Python" != "Python")  # False

are_different = ("apple" != "orange")
print(are_different)  # True
```

### **Less Than: `<`**

### **Greater Than: `>`**

```python
print(10 > 5)    # True
print(5 > 10)    # False
print(5 > 5)     # False

print("zebra" > "apple")  # True (z sau a)
```

## **#7: Working with Operators: Logical**

- Các toán tử như `==, !=, <, >,...` giúp so sánh giá trị và luôn cho ra kết quả là `True` hoặc `False`. Còn **Logical operators** thì dùng để kết hợp nhiều kết quả `True/False` lại với nhau, giúp chúng ta tạo ra những điều kiện phức tạp hơn.

- Thông thường thì một trong các chương trình, chúng ta cần kiểm tra nhiều điều kiện cùng lúc. Vd: muốn biết một người vừa đăng nhập vừa có quyền admin, hoặc hôm nay là thứ Bảy hoặc Chủ Nhật. Lúc này, **Logical operators** sẽ giúp mình làm việc đó. Python có ba **Logical operators** chính: `and, or, not`.

### **The `and` Operator**

- Toán tử `and` kiểm tra hai điều kiện. Nó chỉ trả về `True` khi cả hai điều kiện đều đúng. Nếu chỉ một (hoặc cả hai) sai thì kết quả là `False`.

- Bạn tưởng tượng như muốn vào khu vực an ninh cao: phải có cả thẻ A và thẻ B. Thiếu một cái là không được vào.

```python
age = 25
has_ticket = True

# Is the person old enough AND has a ticket?
can_enter = (age >= 18) and (has_ticket == True)
print(f"Age is 25, has ticket: {can_enter}") # Output: True

age = 16
# Now the first condition (age >= 18) is False
can_enter = (age >= 18) and (has_ticket == True)
print(f"Age is 16, has ticket: {can_enter}") # Output: False

age = 30
has_ticket = False
# Now the second condition (has_ticket == True) is False
can_enter = (age >= 18) and (has_ticket == True)
print(f"Age is 30, no ticket: {can_enter}") # Output: False
```

Dấu ngoặc đơn **parentheses** `()` quanh các phép so sánh không bắt buộc (vì `and` có độ ưu tiên thấp hơn các toán tử so sánh khác), nhưng dùng `()` sẽ giúp code dễ đọc hơn.

### **The `or` Operator**

- Toán tử `or` cũng kiểm tra hai điều kiện, nhưng nó trả về `True` nếu ít nhất một điều kiện đúng. Chỉ khi cả hai đều sai thì mới trả về `False`.

- Tưởng tượng như được giảm giá nếu bạn là học sinh hoặc người cao tuổi. Chỉ cần thỏa một trong hai là đủ.

```python
is_weekend = False
is_holiday = True

# Is it the weekend OR a holiday?
day_off = is_weekend or is_holiday
print(f"Weekend: False, Holiday: True -> Day off: {day_off}") # Output: True

is_weekend = True
is_holiday = False
# The first condition is True
day_off = is_weekend or is_holiday
print(f"Weekend: True, Holiday: False -> Day off: {day_off}") # Output: True

is_weekend = False
is_holiday = False
# Both conditions are False
day_off = is_weekend or is_holiday
print(f"Weekend: False, Holiday: False -> Day off: {day_off}") # Output: False
```

### **The `not` Operator**

- Toán tử `not` đơn giản hơn, nó chỉ làm việc với một điều kiện và đảo ngược kết quả: `True` thành `False`, `False` thành `True`.

- Giống như công tắc đèn: đang bật thì `not` sẽ thành tắt, đang tắt thì `not` sẽ thành bật.

```python
is_logged_in = False
print(f"Is logged in: {is_logged_in}") # Output: False

# Check if the user is NOT logged in
needs_login = not is_logged_in
print(f"Needs login: {needs_login}") # Output: True

is_raining = True
print(f"Is raining: {is_raining}") # Output: True

# Check if it is NOT raining
can_go_outside = not is_raining
print(f"Can go outside: {can_go_outside}") # Output: False
```

### **Combining Logical Operators**

- Bạn có thể ghép nhiều **logical operators** trong cùng một biểu thức. Python sẽ tính `not` trước, rồi đến `and`, cuối cùng mới đến `or`. Nhưng tốt nhất là dùng ngoặc `()` để kiểm soát thứ tự hoặc làm code rõ nghĩa hơn.

```python
age = 22
is_student = True
has_coupon = False

# Eligible if (age < 25 AND is a student) OR has a coupon
is_eligible_discount = (age < 25 and is_student) or has_coupon
# Evaluates as: (True and True) or False
# -> True or False
# -> True
print(f"Eligible for discount: {is_eligible_discount}") # Output: True

# Change coupon status
has_coupon = True
age = 30
is_student = False

is_eligible_discount = (age < 25 and is_student) or has_coupon
# Evaluates as: (False and False) or True
# -> False or True
# -> True
print(f"Eligible for discount (different criteria): {is_eligible_discount}") # Output: True
```

## **#9: Getting User Input**

- Chương trình có thể xử lý dữ liệu được viết sẵn **(hardcode)** ngay trong code, nhưng hầu hết các ứng dụng thực tế đều cần tương tác với người dùng – nghĩa là hỏi thông tin trong lúc chương trình đang chạy. Python cung cấp một cách rất đơn giản để làm việc này: **built-int `input()`**.

- Khi chương trình gọi `input()`, nó sẽ tạm dừng và chờ người dùng gõ gì đó trên bàn phím rồi nhấn Enter. Những gì người dùng gõ sẽ được `input()` trả về dưới dạng chuỗi ký tự (string).

```python
user_name = input("Enter your name: ")
print("Hello,", user_name)
```

### **Input is Always a String**

- Dù người dùng gõ số, chữ, hay bất cứ gì, `input()` luôn luôn trả về một chuỗi. Ví dụ:

```python
age_input = input("Enter your age: ")
print("Type of age_input:", type(age_input))
# Attempting addition might cause an error or unexpected behavior
# next_year = age_input + 1 # This would raise a TypeError
```

### **Converting Input Types**

- Nếu muốn dùng dữ liệu người dùng nhập để tính toán, bạn phải chuyển đổi chuỗi đó sang kiểu phù hợp bằng các hàm đã học: `int()` hoặc `float()`.

```python
age_str = input("Enter your age: ")
age_int = int(age_str) # Convert the input string to an integer

next_year_age = age_int + 1
print("Next year, you will be", next_year_age)

# You can also combine input and conversion in one step:
height_str = input("Enter your height in meters: ")
height_float = float(height_str) # Convert the input string to a float

print("Your height is:", height_float, "meters")
```

- **Lưu ý**: Nếu người dùng gõ lung tung không chuyển được (ví dụ gõ `"hai mươi"` khi bạn dùng `int()`), chương trình sẽ báo lỗi `ValueError`. Cách xử lý lỗi này ta sẽ học sau khi nói về xử lý ngoại lệ **(exception handling)**.

## **#10: Adding Comments to Your Code**

- Các chương trình Python thường có nhiều biến, phép tính phức tạp, hoặc nhiều bước xử lý dữ liệu từ người dùng. Khi code dài lên, việc đọc và hiểu lại code của chính mình (hoặc của người khác) sẽ trở nên khó khăn hơn. Lúc này, việc giữ cho code dễ hiểu là rất quan trọng. Viết code chạy đúng chỉ là một phần; viết code mà sau này (cả bạn và người khác) vẫn dễ đọc, dễ sửa được mới là kỹ năng thực sự tốt. Và **comment** chính là công cụ giúp chúng ta làm điều đó.

- **Comment** là những dòng ghi chú trong code mà Python bỏ qua hoàn toàn khi chạy chương trình. Chúng chỉ tồn tại để giải thích cho con người đọc code hiểu rõ hơn. Bạn có thể tưởng tượng **comment** như những ghi chú nhỏ bạn để lại cho bản thân tương lai hoặc đồng nghiệp.


### **Why Use Comments?**

- **Readability**: **Comment** tốt giúp ai đó nhìn vào code là hiểu ngay đang làm gì. Khi bạn nghỉ vài tuần rồi quay lại code, **comment** sẽ “cứu” bạn rất nhiều.

- **Maintenance**: Khi cần sửa lỗi hoặc thêm tính năng mới, việc hiểu code cũ nhanh chóng là bước đầu tiên. **Comment** giúp tiết kiệm thời gian rất nhiều.

- **Collaboration**: Nếu code của bạn được người khác đọc hoặc chỉnh sửa, **comment** là cách giao tiếp ý đồ và logic rõ ràng nhất.

- **Explaining Complexity**: Có những đoạn code nhìn thì đơn giản nhưng thực ra có lý do đặc biệt để viết như vậy – **comment** sẽ giải thích “tại sao” chứ không chỉ “làm gì”.

### **Writing Comments in Python**

Trong Python, mọi thứ đứng sau dấu `#` trên một dòng đều là **comment** và bị Python bỏ qua.

```python
# This is a full-line comment. It explains the general purpose of the code below.
principal = 1000  # This is an inline comment. It explains this specific variable.
rate = 0.05       # Annual interest rate
years = 3         # Investment duration in years

# Calculate simple interest
interest = principal * rate * years

# Print the result (we'll learn more about formatted printing later)
print("The calculated simple interest is:")
print(interest)
```

### **What Makes a Good Comment?**

- Viết **comment** thì dễ, nhưng viết **comment** hiệu quả mới là quan trọng. Dưới đây là một số nguyên tắc:

  1. **Explain the "Why," Not Just the "What"**:
     - Less helpful: <br>
        `# Assign 5 to x` <br>
        `x = 5`
     - More helpful: <br>
        `# Set the maximum number of retries`<br>
        `max_retries = 5`

  2. **Keep Comments Up-to-Date**: Nếu bạn sửa code, nhớ sửa luôn **comment** tương ứng. **Comment** cũ sai còn tệ hơn không có **comment**.

  3. **Be Concise**: Viết đủ ý, không dài dòng, tránh dùng từ chuyên môn khó hiểu nếu không cần thiết.

  4. **Comment Complex Sections**": Nếu một đoạn logic khó hiểu, có nhiều bước, hoặc dựa trên giả định đặc biệt, hãy thêm **comment** hướng dẫn người đọc.
