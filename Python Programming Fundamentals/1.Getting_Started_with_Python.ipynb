## **#1: What is Python Programming?**

* Programming là quá trình chúng ta viết mã nguồn bằng các ngôn ngữ lập trình, cho phép chúng ta có thể giao tiếp đến máy tính. Về cốt lõi, máy tính chỉ hiểu dãy số `0 và 1`, được gọi là mã máy **(machine code)**. Viết trực tiếp với machine code là vô cùng khó khăn và tốn thời gian. Vì vậy ngôn ngữ lập trình **(programming language)** hoạt động như trung gian, cung cấp một cách dễ đọc và dễ viết hơn, sau cùng sẽ được chuyển thành một dạng mà máy tính có thể thực thi.

* Python là một trong những programming language như vậy. Nó được biết đến như là một ngôn ngữ cấp cao, được thông dịch và có mục đích chung **(high-level, interpreted, general-purpose langauge)**:
  * **High-level**: Điều này có nghĩa là cú pháp của Python được thiết kế để gần với ngôn ngữ con người, nó che giấu đi nhiều các chi tiết phức tạp liên quan đến phần cứng của máy tính.
    * VD: Bạn có thể hình dung giống như khi chỉ đường cho một người lái xe. Thay vì nói chính xác xoay vô-lăng bao nhiêu độ hay đạp ga lúc nào, bạn chỉ cần nói đơn giản:"Rẽ trái ở ngã tư tiếp theo".
    * Nhờ vậy, Python thường dễ đọc và viết và dễ bảo trì hơn so với các ngôn ngữ thấp như C hay Assembly.
  * **Interpreted**: Python thường được thực thi từng dòng bởi một chương trình gọi là trình thông dịch **(interpreter)**. Khi bạn chạy một tệp lệnh Python, **interpreter** sẽ đọc mã của bạn và thực hiện các hành động được chỉ định ngay lập tức. Điều này trái ngược với các ngôn ngữ biên dịch **(compiled langauges)**, toàn bộ chương trình phải được chuyển sang mã máy trước, rồi mới có thể chạy được.
  * **General-Purpose**: Python không được thiết kế chỉ cho một nhiệm vụ cụ thể. Nó linh hoạt và được sử dụng trên nhiều ứng dụng bao gồm web, data analysis, AI,v.v..

### **Core Characteristics of Python - Đặc điểm cốt lõi của Python**
* Dễ đọc và đơn giản **(readability and simplicity)**: Như đề cập ở trên, cú pháp rõ ràng và giúp bạn dễ tìm hiểu các nguyên tắc cơ bản lập trình mà không bị lún sâu vào các quy tắc phức tạp.

* Thư viện tiêu chuẩn rộng lớn **(Large Standard Library)**: Python đi kèm với một bộ sưu tập phong phú các **modules** có sẵn, xử lý các tác vụ phổ biến như làm việc với `text, numbers, files, networking, ...`. Điều này có nghĩa là bạn thường sẽ không cần phải viết lại mã từ đầu cho các thao tác cơ bản này.

* Cộng đồng và hệ sinh thái sôi động **(Vibrant Community and Ecosystem)**:Python được hưởng lợi từ một cộng đồng lớn toàn cầu. Điều này giúp nguồn tài liệu học tập phong phú, diễn đàn trợ giúp và kho lưu trữ khổng lồ các gói thư viện của bên thứ ba có sẵn thông qua các công cụ như `pip`.

* **Dynamic Typing**: Trong Python, bạn sẽ không cần thiết phải khai báo rõ ràng kiểu dữ kiểu **(type of data)** cho **variable** sẽ chứa **value**. Python tự động tìm ra điều này khi chương trình của bạn được **interpret**. Mặc dù điều này mang lại sự linh hoạt, nhưng việc hiểu **type of data** vẫn rất quan trọng.
