##  **#1: Strong Information: Variables**



### **What is a Variable?**

Trong Python, **variable** là một tên tượng trưng (nhãn) được dùng để tham chiếu đến một vị trí trong bộ nhớ máy tính, nơi lưu trữ các giá trị (dữ liệu). Chúng giống như những gói hàng có nhãn, giúp lập trình viên dễ dàng quản lý, truy cập và thao tác với dữ liệu (int, string, list..).

Hãy tưởng tượng bạn đang sắp xếp các mặt hàng trong hộp. Bạn sẽ không chỉ đặt mọi thứ vào những hộp không được đánh dấu, bạn sẽ gắn nhãn cho chúng ('Quần áo mùa đông', 'sách',...). Các **variable** phục vụ mục đích tương tự trong lập trình: Gán tên mô tả cho các bộ nhớ chứa dữ liệu.


### **Creating Variables: The Assignment Statement**

Việc tạo một **variable** trong Python vô cùng đơn giản sử dụng **asignment operator**, ký hiệu là `=`:

`variable_name = value`

```python
message = 'Hello, Python'
# message is a variable storing a string of text
count = 10
# count is a variable storing an integer
price = 49.99
# price is a variable storing a floating-point number
is_active = True
# is_active is a variable storing a boolean value
```

Vì Python là **dynamically typed**, có nghĩa là bạn không cần phải khai báo rõ ràng **type of data** cho **variable**. Python sẽ tự động xác định type dựa trên **value** mà bạn gán cho nó.

### **Using Variables**

* Lưu trữ và quản lý: Giúp lưu giữ thông tin cần thiết trong suốt chương trình.

* Dễ đọc và hiểu: Dùng tên **variable** mô tả (ví dụ ten_nguoi_dung) thay vì giá trị thô (ví dụ "An") làm code dễ hiểu hơn.

* Linh hoạt: Cho phép thay đổi giá trị dễ dàng; ví dụ, thay đổi thuế suất chỉ cần sửa giá trị **variable** `tax`.

* Tái sử dụng: Dùng lại giá trị nhiều lần trong code mà không cần viết lại

```python
message = ''
print(message) # Output:

items = 5
cost_per_item = 2.5
total_cost = items * cost_per_item # use variables in calculations
print(total_cost) # Output: 12.5
```

### **Naming Variables: Rules and Conventions**

Việc đặt tên **variable** là rất quan trọng để đánh giá lập trình viên, nó giúp người đọc code dễ hiểu và dễ bảo trì. Python đã cung cấp các quy tắc và quy ước nhằm cho **naming variables**:

* **Rules**:
  * **variable** phải bắt đầu bằng chửi cái `(a-z, A-Z)`, hoặc là dấu gạch dưới `_`, phần còn lại thể chứa các chữ cái, số `(0-9)`
  * **variable** phân biệt chữ hoa chữ thường. `myVariable, myvariable và MYVARIABLE`  là các 3 **variables** riêng biệt.
  * **variable** không được giống với [Python keywords](https://www.w3schools.com/python/python_ref_keywords.asp).

* **Conventions**:
 * Hãy dùng những tên **variable** rõ ràng và có ý nghĩa, thể hiện đúng mục đích của **variable**.
   * VD: Nên viết `user_name` thay vì `un`, hoặc `total_price` thay vì `tp`.
 * Khi tên **variable** gồm nhiều từ, hãy dùng `snake_case`.Điều này có nghĩa là tất cả các chữ đều viết thường và các từ được nối với nhau bằng dấu gạch dưới `_`.
   * VD: `first_name` hoặc `number_of_items`
 * Tránh dùng các tên **variable** chỉ có một ký tự như `l, O hoặc I` vì chúng rất dễ bị nhầm với số `1 và 0`. Các tên ngắn như `x, y hoặc i` có thể chấp nhận được trong một số trường hợp đặc biệt, chẳng hạn như **variable** đếm trong vòng lặp hoặc tọa độ toán học, nhưng nhìn chung vẫn nên ưu tiên các tên **variable** mang tính mô tả.


```python
myvar = "John_1"
my_var = "John_2"
_my_var = "John_3"
myVar = "John_4"
MYVAR = "John_5"
myvar2 = "John_6"

print(myvar)
print(my_var)
print(_my_var)
print(myVar)
print(MYVAR)
print(myvar2)

#This example will produce an error in the result

2myvar = "John_7"
my-var = "John_8"
my var = "John_9"
```

### **Changing Variable Values (Reassignment)**

Ta có thể thay đổi giá trị được lưu trữ trong **variable** bằng cách gán lại giá trị mới cho **variable** đó bằng toán tử gán `=`.

```python
# Assign an initial value
score = 100
print("Initial score:", score) # Output: Initial score: 100

# Update the value
score = 150
print("Updated score:", score) # Output: Updated score: 150

# Assign a value based on the current value
score = score + 20 # Read the current value (150), add 20, store the result (170) back
print("Final score:", score)   # Output: Final score: 170
```

##  **#2: Numbers (Integers, Floats)**


### **Integers (int)**

**Integers** đại diện cho số nguyên trong python. Chúng có thể là dương, âm hoặc 0, nhưng chúng không thể là phân số hoặc thập phân.

**Integers** có thể lớn tùy ý, nó chỉ bị giới hạn bởi bộ nhớ của hệ thống. Bạn không cần phải lo lắng về `size` như `short`, `int`, `long` ở các ngôn ngữ lập trình khác.

Bạn cũng có thể kiểm trả **data types** với `type()`

```python
# Examples of integers
user_count = 150
temperature_celsius = -5
year = 2024
zero_value = 0

print(user_count)
print(temperature_celsius)
print(type(year))
print(type(zero_value))
```

### **Floating-Point Numbers (float)**

**Floating-point numbers** hay **float** là kiểu dữ liệu dùng để biểu diễn các số thập phân (ví dụ: 3.14, -0.5, 2.0).

```python
# Examples of floating-point numbers
pi_approx = 3.14159
price = 49.99
temperature_fahrenheit = -20.5
account_balance = 0.0 # Using a decimal makes it a float
distance_meters = 1.5e3 # Scientific notation for 1.5 * 10^3, which is 1500.0

print(pi_approx)
print(distance_meters)

print(type(price))
print(type(temperature_fahrenheit))
print(type(account_balance))
```

### **0.1 + 0.2 = ?**

- Điều quan trong cần lưu ý là **floating-point numbers** trên máy tính không phải lúc nào cũng hoàn toàn chính xác. Do cách lưu trữ **floating-point numbers** nội bộ, thường lưu ở định dạng `nhị phân IEEE 754`, các phép tính liên quan đến **float** đôi khi có thể mang lại kết quả rất khác so với những gì bạn mong đợi về mặt toán học.

  -  Vd: `0.1 + 0.2` có thể dẫn đến kết quả `0.30000000000000004` thay vì `0.3`

- Trong hầu hết các trường hợp, sự sai lệch rất nhỏ này không gây ảnh hưởng gì đáng kể. Tuy nhiên, bạn cần chú ý trong các bài toán tài chính hoặc mô phỏng khoa học, nơi yêu cầu độ chính xác cao.

- Để khắc phục vấn đề này, python cung cấp `Decimal` để xử lý các trường hợp tương tự.

```python
result = 0.1 + 0.2
print(result) # output: 0.30000000000000004
```

```python
from decimal import Decimal

result = Decimal('0.1') + Decimal('0.2')
print(result) # output: 0.3
```

## **#3: Text (Strings)**

- **Text (String)** là kiểu dữ liệu cơ bản dùng để biểu diễn văn bản, là một dãy các ký tự (chữ cái, số, ký tự đặc biệt) được đặt trong cặp dấu nháy đơn `( ' ' )`, nháy kép `( " " )` hoặc ba dấu nháy `( ''' ''' hoặc """ """ )`.

```python
# Using single quotes
message_single = 'Hello, Python learner!'
print(message_single)

# Using double quotes
message_double = "This also works just fine."
print(message_double)

# What if your string contains a single quote? Use double quotes to enclose it.
quote = "He said, 'Python is fun!'"
print(quote)

# What if your string contains a double quote? Use single quotes.
reply = 'She responded, "Indeed it is."'
print(reply)

# Triple double quotes for multi-line string
multi_line_doc = """This is the first line.
This is the second line.
And "quotes" or 'apostrophes' can be used freely inside."""
print(multi_line_doc)

# Triple single quotes work the same way
multi_line_alt = '''Another way to write
multi-line text,
very convenient.'''
print(multi_line_alt)
```

### **Basic String Operations**
Bạn có thể kết hợp các chuỗi **(combine strings)** bằng toán tử **concatenation `+`** và **repetition `*`**

```python
first_name = "Ada"
last_name = "Lovelace"
space = " "

# Concatenation
full_name = first_name + space + last_name
print(full_name) # Output: Ada Lovelace

# Repetition
separator = "-" * 10 # Repeat the hyphen 10 times
print(separator)   # Output: ----------
print(first_name * 3) # Output: AdaAdaAda
```

Lưu ý về bạn **concatenation** string này trực tiếp với con những con số, thì chương trình sẽ bị lỗi. Trước tiên, bạn cần chuyển đổi số thành một string, sử dụng str(), mà chúng ta sẽ sớm đề cập trong **Type Conversion**.

### **Accessing Characters: Indexing**

Vì string là những sequences, bạn có thể truy cập các ký tự riêng lẽ bẳng cách sử dụng chỉ mục **(index)** của chúng. Python được thiết lập **index** bắt đầu từ con số `0`, nghĩa là `index` đầu tiên sẽ là `0`.

```python
language = "Python"

# Accessing characters by positive index
print(language[0]) # Output: P (the first character)
print(language[1]) # Output: y (the second character)
print(language[5]) # Output: n (the sixth character)

# Accessing characters using negative index (starts from the end)
print(language[-1]) # Output: n (the last character)
print(language[-2]) # Output: o (the second-to-last character)
```

```python
language = "Python"

# Trying to access an index that doesn't exist will result in an IndexError.
print(language[6])
```

### **Extracting Substrings: Slicing**

Nếu bạn cần lấy nhiều hơn một ký tự, bạn có thể trích xuất một phần của string, gọi là **substring** hoặc **slice**.

Thao tác này được gọi là **slicing** và sử dụng cú pháp `[start:stop:step]`.

- `string[start : end : step]`

  - `start`: vị trí bắt đầu của **slice**.  Nếu bỏ qua, mặc định là `0`.
  - `end`: Vị trí kết thúc của **slice**. Nếu bỏ qua, mặc định là đến cuối string.
  - `step`: Bước nhảy (số ký tự bỏ qua). Nếu bỏ qua, mặc định là `1`.

```python
language = "Python"

# Get characters from index 1 up to (but not including) index 4
print(language[1:4]) # Output: yth

# Get characters from the beginning up to index 3 (exclusive)
print(language[:3])  # Output: Pyt

# Get characters from index 2 to the end
print(language[2:])  # Output: thon

# Get the whole string (less common, but shows the defaults)
print(language[:])   # Output: Python

# Get every second character
print(language[::2]) # Output: Pto

# Get the string in reverse
print(language[::-1]) # Output: nohtyP
```

**Slicing** luôn tạo ra một string mới và không bao giờ gây ra lỗi `IndexError`, ngay cả khi các chỉ số bắt đầu hoặc kết thúc vượt quá độ dài của string.
Thay vào đó, Python sẽ chỉ trả về phần string có thể lấy được trong phạm vi được yêu cầu.

```python
s = "Python"

print(s[0:100])   # Python
print(s[3:50])    # hon
print(s[-100:3])  # Pyt
```

### **String Immutability**

- Một đặc điểm quan trọng của string trong Python là chúng bất biến **(immutable)**.
Điều này có nghĩa là một khi string đã được tạo ra, bạn không thể thay đổi trực tiếp nội dung của nó.

- Những thao tác trông giống như đang sửa string, chẳng hạn như **concatenation** và **repetition** hoặc các phương thức xử lý string, thực chất là tạo ra một string mới và trả về string đó, chứ không làm thay đổi string ban đầu.

```python
greeting = "hello"
# greeting[0] = "H" # This will cause a TypeError: 'str' object does not support item assignment

# To "change" the greeting, create a new string
concatenation = "H" + greeting[1:]
repetition = "H" * 3
print(concatenation) # Output: Hello
print(repetition)     # Output: hello (original string is unchanged)
```

### **Formatting Strings**

- Thông thường, bạn sẽ cần tạo các string có chứa giá trị của biến.
Mặc dù có thể dùng **concatenation** (+), cách này nhanh chóng trở nên rườm rà và dễ gây lỗi, đặc biệt khi làm việc với các kiểu dữ liệu không phải là string.

- Python cung cấp những cách tốt hơn và gọn gàng hơn để định dạng string.
Cách hiện đại và được khuyến khích nhất là sử dụng `f-string`, được giới thiệu từ Python 3.6.

- Để tạo `f-string`, bạn chỉ cần thêm chữ `f` hoặc `F` trước string.
Bên trong string, bạn có thể chèn trực tiếp biến hoặc biểu thức bằng cách đặt chúng trong dấu ngoặc nhọn `{}`.



```python
name = "Alice"
age = 30
city = "New York"

# Using f-string for clean formatting
intro_fstring = f"My name is {name}, I am {age} years old, and I live in {city}."
print(intro_fstring)
# Output: My name is Alice, I am 30 years old, and I live in New York.

# You can also include expressions inside the braces
radius = 5
area = 3.14159 * radius**2
print(f"A circle with radius {radius} has an area of {area:.2f}.")
# Output: A circle with radius 5 has an area of 31.42.
# Note: :.2f formats the float to 2 decimal places.
```

- `f-string` ngắn gọn và giúp ta dễ đọc code, nó được sử dụng để thay thể các phương thức format cũ khác như `str.format()` hoặc toán tử `%` ở các phiên bản Python cũ

```python
name = "Alice"
age = 25

print("My name is %s and I am %d years old." % (name, age)) # %s -> string, %d -> int
print("My name is {} and I am {} years old.".format(name, age))
```

### **Escape Sequences**

Đôi khi, bạn cần đưa vào string những ký tự đặc biệt - những ký tự khó hoặc không thể gõ trực tiếp, hoặc những ký tự đã có ý nghĩa đặc biệt trong Python.
Ví dụ, bạn muốn chèn ký tự xuống dòng, tab, hoặc thêm dấu ngoặc kép vào bên trong một string đang dùng chính loại ngoặc đó.

Trong Python, việc này được thực hiện bằng **escape sequence**, bắt đầu bằng dấu gạch chéo ngược `\`.

- Các **escape sequences** thường gặp:
| Escape | Ý nghĩa              |
| ------ | -------------------- |
| `\n`   | Xuống dòng (newline) |
| `\t`   | Tab                  |
| `\\`   | Dấu `\`              |
| `\'`   | Dấu nháy đơn `'`     |
| `\"`   | Dấu nháy kép `"`     |

```python
# Newline
print("First line.\nSecond line.")
# Output:
# First line.
# Second line.

# Tab
print("Column1\tColumn2")
# Output: Column1    Column2

# Literal backslash
print("This is a path: C:\\Users\\Name")
# Output: This is a path: C:\Users\Name

# Literal quotes within string defined by same quotes
print('He said, \'Hello!\'') # Output: He said, 'Hello!'
print("She replied, \"Hi!\"") # Output: She replied, "Hi!"
```

### **Common String Methods**

- String trong Python có rất nhiều phương thức **(methods)** dựng sẵn để thực hiện các thao tác phổ biến.

- **Method** giống như hàm **(function)**, nhưng được gọi trực tiếp trên một đối tượng bằng `.`.Ở đây, đối tượng chính là string.

Dưới đây là một số **functions** và **method** string thường dùng:

| Method / Function | Mô tả (English)               | Mô tả (Tiếng Việt)          | Ví dụ                      |
| ----------------- | ----------------------------- | --------------------------- | -------------------------- |
| `len(string)`          | Returns number of characters  | Trả về số ký tự trong chuỗi | `len("Hello") → 5`         |
| `s.lower()`       | Convert to lowercase          | Chuyển sang chữ thường      | `"Hi".lower() → "hi"`      |
| `s.upper()`       | Convert to uppercase          | Chuyển sang chữ hoa         | `"hi".upper() → "HI"`      |
| `s.strip()`       | Remove whitespace (both ends) | Xóa khoảng trắng đầu & cuối | `" hi ".strip() → "hi"`    |
| `s.lstrip()`      | Remove left whitespace        | Xóa khoảng trắng bên trái   | `" hi".lstrip()`           |
| `s.rstrip()`      | Remove right whitespace       | Xóa khoảng trắng bên phải   | `"hi ".rstrip()`           |
| `s.find(substring)`     | Find substring index          | Tìm vị trí chuỗi con        | `"hello".find("e") → 1`    |
| `s.replace(a, b)` | Replace substring             | Thay chuỗi con              | `"a-b".replace("-", "_")`  |
| `s.startswith(prefix)` | Check prefix                  | Kiểm tra bắt đầu bằng       | `"hi.py".startswith("hi")` |
| `s.endswith(suffix)`   | Check suffix                  | Kiểm tra kết thúc bằng      | `"hi.py".endswith(".py")`  |
| `s.split(separator)`    | Split into list               | Tách chuỗi thành list       | `"a,b".split(",")`         |
| `sep.join(iterable)`    | Join iterable                 | Nối list thành chuỗi        | `" ".join(["I","AI"])`     |


```python
text = "  Learning Python is Fun!  "

# Length
print(len(text)) # Output: 27

# Case conversion
print(text.lower()) # Output:   learning python is fun!
print(text.upper()) # Output:   LEARNING PYTHON IS FUN!

# Stripping whitespace
print(text.strip())  # Output: Learning Python is Fun!
print(text.lstrip()) # Output: Learning Python is Fun!
print(text.rstrip()) # Output:   Learning Python is Fun!

# Finding substrings
print(text.find("Python")) # Output: 11 (index where 'Python' starts)
print(text.find("Java"))   # Output: -1 (not found)

# Replacing substrings
print(text.replace("Fun", "Awesome")) # Output:   Learning Python is Awesome!

# Checking start/end (use strip first for accuracy here)
clean_text = text.strip()
print(clean_text.startswith("Learning")) # Output: True
print(clean_text.endswith("!"))        # Output: True

# Splitting a string into a list
words = clean_text.split(" ")
print(words) # Output: ['Learning', 'Python', 'is', 'Fun!']

# Joining a list into a string
joined_string = "---".join(words)
print(joined_string) # Output: Learning---Python---is---Fun!

# Original string remains unchanged due to immutability
print(text) # Output:   Learning Python is Fun!
```
